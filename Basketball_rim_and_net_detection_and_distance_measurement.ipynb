```python
import torch
import cv2
import numpy as np
from models.experimental import attempt_load
from utils.general import non_max_suppression, scale_boxes
from utils.augmentations import letterbox

# Configuration
WEIGHTS_PATH = "runs/train/exp/weights/best.pt"  # Adjust 'exp' if needed
IMG_SIZE = 640  # Image size for inference
DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"  # Use GPU if available
CONF_THRES = 0.25  # Confidence threshold
IOU_THRES = 0.45  # NMS IoU threshold
CAMERA_ID = 0  # Adjust based on your webcam ID (from Step 1)

# Load the YOLOv5 model
model = attempt_load(WEIGHTS_PATH, device=DEVICE)
model.eval()
stride = int(model.stride.max())  # Model stride
names = model.names  # Class names (e.g., 'rim', 'net')

# Initialize webcam
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print(f"Error: Could not open webcam with ID {CAMERA_ID}")
    exit()

# Set window name
cv2.namedWindow("Basketball Rim Detection", cv2.WINDOW_NORMAL)

print("Starting real-time detection... Press 'q' to quit.")

while True:
    # Read frame from webcam
    ret, frame = cap.read()
    if not ret:
        print("Error: Failed to capture frame from webcam")
        break

    # Preprocess the frame
    img = letterbox(frame, new_shape=IMG_SIZE, stride=stride, auto=True)[0]  # Resize with padding
    img = img.transpose(2, 0, 1)  # HWC to CHW
    img = np.ascontiguousarray(img)
    img = torch.from_numpy(img).to(DEVICE).float() / 255.0  # Normalize
    img = img.unsqueeze(0)  # Add batch dimension

    # Run inference
    with torch.no_grad():
        pred = model(img)[0]
    pred = non_max_suppression(pred, CONF_THRES, IOU_THRES)

    # Process detections
    for det in pred:  # Detections per frame
        if len(det):
            # Rescale boxes from img_size to original frame size
            det[:, :4] = scale_boxes(img.shape[2:], det[:, :4], frame.shape).round()
            # Draw bounding boxes and labels
            for *xyxy, conf, cls in det:
                if names[int(cls)] == "rim":  # Only draw boxes for 'rim' class
                    label = f"{names[int(cls)]} {conf:.2f}"
                    cv2.rectangle(frame, (int(xyxy[0]), int(xyxy[1])), (int(xyxy[2]), int(xyxy[3])), (0, 255, 0), 2)
                    cv2.putText(frame, label, (int(xyxy[0]), int(xyxy[1]) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # Display the frame
    cv2.imshow("Basketball Rim Detection", frame)

    # Break the loop if 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
print("Real-time detection stopped.")
```

```python
import torch
import cv2
import numpy as np
from models.experimental import attempt_load
from utils.general import non_max_suppression, scale_boxes
from utils.augmentations import letterbox

# Configuration
WEIGHTS_PATH = "runs/train/exp/weights/best.pt"
VIDEO_PATH = "C:/Projects/basketball_project/test_videos/rim_video.mp4"
OUTPUT_VIDEO_PATH = "C:/Projects/basketball_project/test_videos/rim_video_detected.mp4"
IMG_SIZE = 640
DEVICE = "cuda:0" if torch.cuda.is_available() else "cpu"
CONF_THRES = 0.25
IOU_THRES = 0.45

# Distance estimation parameters
REAL_RIM_HEIGHT = 0.457  # Standard rim diameter in meters (18 inches)
CALIBRATION_DISTANCE = 5.0  # Known distance in meters for calibration
CALIBRATION_PIXEL_HEIGHT = None  # To be determined during calibration (run the script once to find this)

# Load the YOLOv5 model
model = attempt_load(WEIGHTS_PATH, device=DEVICE)
model.eval()
stride = int(model.stride.max())
names = model.names

# Load the video
cap = cv2.VideoCapture(VIDEO_PATH)
if not cap.isOpened():
    print(f"Error: Could not open video at {VIDEO_PATH}")
    exit()

# Get video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS)) or 30
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
print(f"Video resolution: {width}x{height} at {fps} FPS, {total_frames} frames")

# Initialize video writer
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(OUTPUT_VIDEO_PATH, fourcc, fps, (width, height))

# Set window name
cv2.namedWindow("Basketball Rim Detection", cv2.WINDOW_NORMAL)
cv2.resizeWindow("Basketball Rim Detection", width, height)

print("Processing video... Press 'q' to quit early.")

frame_count = 0
scaling_factor = None  # To be calculated after calibration

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print("Finished processing video or error reading frame.")
        break

    frame_count += 1
    print(f"Processing frame {frame_count}/{total_frames}", end="\r")

    # Preprocess the frame
    img = letterbox(frame, new_shape=IMG_SIZE, stride=stride, auto=True)[0]
    img = img.transpose(2, 0, 1)
    img = np.ascontiguousarray(img)
    img = torch.from_numpy(img).to(DEVICE).float() / 255.0
    img = img.unsqueeze(0)

    # Run inference
    with torch.no_grad():
        pred = model(img)[0]
    pred = non_max_suppression(pred, CONF_THRES, IOU_THRES)

    # Process detections
    rim_detected = False
    for det in pred:
        if len(det):
            det[:, :4] = scale_boxes(img.shape[2:], det[:, :4], frame.shape).round()
            for *xyxy, conf, cls in det:
                if names[int(cls)] == "rim":
                    rim_detected = True
                    # Draw bounding box and label
                    label = f"{names[int(cls)]} {conf:.2f}"
                    cv2.rectangle(frame, (int(xyxy[0]), int(xyxy[1])), (int(xyxy[2]), int(xyxy[3])), (0, 255, 0), 2)
                    cv2.putText(frame, label, (int(xyxy[0]), int(xyxy[1]) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

                    # Calculate pixel height of the rim
                    pixel_height = int(xyxy[3]) - int(xyxy[1])

                    # Calibration step (run this once at a known distance)
                    if CALIBRATION_PIXEL_HEIGHT is None and frame_count == 1:
                        CALIBRATION_PIXEL_HEIGHT = pixel_height
                        scaling_factor = CALIBRATION_DISTANCE * CALIBRATION_PIXEL_HEIGHT / REAL_RIM_HEIGHT
                        print(f"Calibration: Pixel height at {CALIBRATION_DISTANCE}m = {CALIBRATION_PIXEL_HEIGHT}px")
                        print(f"Scaling factor: {scaling_factor}")

                    # Estimate distance
                    if scaling_factor is not None and pixel_height > 0:
                        distance = (REAL_RIM_HEIGHT * scaling_factor) / pixel_height
                        distance_label = f"Distance: {distance:.2f}m"
                        cv2.putText(frame, distance_label, (int(xyxy[0]), int(xyxy[1]) - 30), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 2)

    # If no rim is detected, display a message
    if not rim_detected:
        cv2.putText(frame, "No rim detected", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

    # Write the frame to the output video
    out.write(frame)

    # Display the frame
    cv2.imshow("Basketball Rim Detection", frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()
print(f"\nVideo processing complete. Output saved to {OUTPUT_VIDEO_PATH}")
print(f"Calibrated pixel height at {CALIBRATION_DISTANCE}m: {CALIBRATION_PIXEL_HEIGHT}px")
```
